````python
# Вопросы 2 и 3

PERSON_FIELDS = [
    "id",
    "name",
    "gender",
    "birth_year",
    "country_birth",
    "country_now",
    "prizes_relevant",
    "type_",
]

ORG_FIELDS = [
    "id",
    "name",
    "founded_year",
    "country_founded",
    "country_now",
    "prizes_relevant",
    "type_",
]


def count_missing_values(data, fields):
    missing = {field : 0 for field in fields}

    for record in data:
        for field in fields:
                if (
                    field not in record
                    or record.get(field) is None
                    or record.get(field) == ""
                ):
                     missing[field] += 1
    return missing


def calculate_missing_shares(
    missing_values,
    total_records,
):
    if total_records == 0:
        print("Total number of records is 0")
        return None
    missing_shares = {
         key : round(value / total_records , 3)
         for key, value in missing_values.items()
    }
    return missing_shares

missing_values_persons = count_missing_values(persons, PERSON_FIELDS)
missing_shares_persons = calculate_missing_shares(
     missing_values_persons,
     len(persons),
)

missing_values_organizations = count_missing_values(organizations, ORG_FIELDS)
missing_shares_organizations = calculate_missing_shares(
     missing_values_organizations,
     len(organizations),
)

print("Пропущенные поля у лауреатов-людей:")
print(missing_values_persons)
print("Доля пропущенных полей у лауреатов-людей:")
print(missing_shares_persons)
print("Пропущенные поля у лауреатов-орагинзаций:")
print(missing_values_organizations)
print("Доля пропущенных полей у лауреатов-орагинзаций:")
print(missing_shares_organizations)
````

````python
from aggregations import (
    mean_value,
    median_value,
    top_n_counts,
    count_by_value
)
from dataset_helpers import (
    group_by_attributes,
    aggregate_groups,
    flatten_nested_list,
    filter_by_lambda,
)
````

````python
# 2.1. Топ-статистика по странам

print("Топ-5 стран по общему количеству лауреатов:")
for country, count in top_n_counts(persons, "country_birth", 5):
    print(f"\t{country}: {count}")

flattened_persons = flatten_nested_list(persons, "prizes_relevant")

# Каждый ключ — уникальное сочетание ID, страны рождения и категории.
uniq_prizes_by_country_category = group_by_attributes(
    flattened_persons,
    ["id", "country_birth", "category_en"],
).keys()

uniq_categories = Counter(
    x[2] for x in uniq_prizes_by_country_category
)

for category, _ in uniq_categories.most_common():
    countries = [
        country
        for _, country, prize_category in uniq_prizes_by_country_category
        if prize_category == category and country is not None
    ]

    print(f"\nТоп-5 стран в категории {category}:")
    for country, count in Counter(countries).most_common(5):
        print(f"\t{country}: {count}")
````

````python
# **2.2. Возраст при награждении**
# - Рассчитать средний возраст лауреатов на момент получения **первого** приза.
#  Рассмотрите только лауреатов-людей.
# - Исследуйте, как средний возраст получения первого приза различается
#  по категориям
# - Различаются ли средний и медианный возрасты?


def get_first_award_age(person, laureates_list):
    award_years = get_awards_year_by_id(laureates_list, person["id"])
    first_award_year = min(award_years)

    first_prize = [
        prize for prize in person["prizes_relevant"]
        if prize["award_year"] == first_award_year
    ][0]

    return {
        "name": person["name"],
        "category_en": first_prize["category_en"],
        "age_at_first_award": first_award_year - person["birth_year"],
    }


age_data = [
    get_first_award_age(person, laureates_data)
    for person in persons
]

# Средний и медианный возраст по всем людям.
mean_age = mean_value(age_data, "age_at_first_award")
median_age = median_value(age_data, "age_at_first_award")

print(f"Средний возраст: {mean_age:.2f}")
print(f"Медианный возраст: {median_age:.2f}")
print(f"Разница между средним и медианой: {abs(mean_age - median_age):.2f}")

# Средний возраст по категории первой премии.
age_by_category = group_by_attributes(age_data, ["category_en"])
mean_age_by_category = aggregate_groups(
    age_by_category,
    mean_value,
    "age_at_first_award",
)

print("Средний возраст по категориям:")
mean_categ_age_data = sorted(
    mean_age_by_category.items(),
    key=lambda item: item[1],
    reverse=True,
)
for (category,), age in mean_categ_age_data:
    print(f"\t{category}: {age:.2f}")
````

````python
# 2.6. Отказы от премий

prize_records = flatten_nested_list(laureates_data, "prizes_relevant")

declined_prizes = filter_by_lambda(
    prize_records,
    lambda record: record["prize_status"] == "declined",
)

declines_by_category = count_by_value(declined_prizes, "category_en")

print("Количество отказов по категориям:")
for category, count in declines_by_category.most_common():
    print(f"\t{category}: {count}")

organization_declines = filter_by_lambda(
    declined_prizes,
    lambda record: record["type_"] == "org",
)

print(f"\nОтказов среди организаций: {len(organization_declines)}")
````
